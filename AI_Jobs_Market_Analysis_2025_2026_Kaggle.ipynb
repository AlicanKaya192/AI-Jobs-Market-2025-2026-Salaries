```python
# Search recursively for the CSV dataset file across common locations
csv_search_patterns = [
    '**/ai_jobs_market_2025_2026.csv',
    '../**/ai_jobs_market_2025_2026.csv',
    '/kaggle/input/**/ai_jobs_market_2025_2026.csv'
]

dataset_path = None
for pattern in csv_search_patterns:
    matched_files = glob.glob(pattern, recursive=True)
    for path in matched_files:
        if os.path.exists(path) and os.path.isfile(path):
            dataset_path = path
            break
    if dataset_path:
        break

if dataset_path is None:
    # Local fallback
    dataset_path = 'dataset/ai_jobs_market_2025_2026.csv'

df = pd.read_csv(dataset_path)
print(f"Successfully loaded {len(df)} postings from: {dataset_path}")
```

## 5. 🔍 Step-by-Step Data Analysis & Visualizations

<a id='5.1'></a>
### 5.1 🏆 Highest Paying AI Job Titles

#### What This Code Does:
This script groups all job postings by their exact job title and calculates:
- The **median annual salary** (to avoid outlier distortions).
- The **average demand score** out of 100.
- The **total volume (count)** of job postings for each title.
It then extracts the top 15 highest-paying titles and draws a horizontal bar chart.

#### Why We Wrote It This Way:
- **Horizontal Bars**: A horizontal layout allows long job titles to be read clearly on the Y-axis.
- **Annotations**: Adding both the median salary and the mean demand score next to each bar directly links compensation with market demand.
- **Formatting**: `mticker.FuncFormatter` is used to scale the values into standard `$k` formatting for clean aesthetics.


```python
title_stats = (df.groupby('job_title')
                 .agg(median_salary=('annual_salary_usd','median'),
                      demand=('demand_score','mean'),
                      count=('job_id','count'))
                 .reset_index()
                 .sort_values('median_salary', ascending=False)
                 .head(15))

fig, ax = plt.subplots(figsize=(13, 8))
fig.patch.set_facecolor('#FAFAF8')
ax.set_facecolor('#FAFAF8')

colors = [PALETTE[i % len(PALETTE)] for i in range(len(title_stats))]
# Reverse with [::-1] so the highest paying job is at the top
bars = ax.barh(title_stats['job_title'][::-1],
               title_stats['median_salary'][::-1] / 1000,
               color=colors[::-1], alpha=0.88,
               edgecolor='white', linewidth=0.8)

# Text annotation on the side of each bar
for bar, (_, row) in zip(bars, title_stats[::-1].iterrows()):
    ax.text(bar.get_width() + 1,
            bar.get_y() + bar.get_height() / 2,
            f'${row["median_salary"]/1000:.0f}k  |  demand: {row["demand"]:.0f}/100',
            va='center', fontsize=8.5, color='#333', fontweight='bold')

ax.set_xlabel('Median Annual Salary (USD thousands)', fontsize=11)
ax.set_title('Highest Paying AI Job Titles in 2025–2026\nMedian Salary + Demand Score',
             fontsize=14, fontweight='bold', pad=15)
ax.set_xlim(0, 420)
ax.xaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'${x:.0f}k'))
ax.text(0.01, -0.06,
        f'Source: AI Jobs Market 2025–2026 Dataset | n={len(df)} job postings',
        transform=ax.transAxes, fontsize=7.5, color='#888', style='italic')

plt.tight_layout()
plt.savefig('results/ai_top_paying_job_titles.png', dpi=180, bbox_inches='tight', facecolor='#FAFAF8')
plt.show()
```

<a id='5.2'></a>
### 5.2 📈 Salary Distribution by Experience Level

#### What This Code Does:
This section maps AI salaries to four experience levels: Entry, Mid, Senior, and Lead. It renders:
- A **Box Plot**: Showing the median, 25th-75th percentiles (IQR), and range.
- A **Jittered Scatter Overlay**: Randomly spreading the original data points horizontally so we can visualize density and volume.
- **Step-up Badges (Deltas)**: Computing and drawing the difference between the median salaries of consecutive experience levels.

#### Why We Wrote It This Way:
- **Box plots alone can hide density**: Overlaying jittered scatter plots ensures that we can see exactly where the bulk of the listings sit.
- **Deltas highlight growth**: By calculating the difference (`+$Xk`) between consecutive levels and rendering it inside colored badges, we make the financial return of experience instantly clear to career planners.


```python
exp_order = ['Entry (0-2 yrs)', 'Mid (3-5 yrs)', 'Senior (6-9 yrs)', 'Lead (10+ yrs)']
exp_data  = [df[df['experience_level'] == e]['annual_salary_usd'].values for e in exp_order]
exp_colors = ['#4CAF50', '#2196F3', '#FF9800', '#9C27B0']
exp_labels = ['Entry\n(0–2 yrs)', 'Mid\n(3–5 yrs)', 'Senior\n(6–9 yrs)', 'Lead\n(10+ yrs)']

fig, ax = plt.subplots(figsize=(11, 7))
fig.patch.set_facecolor('#FAFAF8')
ax.set_facecolor('#FAFAF8')

bp = ax.boxplot(exp_data, patch_artist=True,
                medianprops=dict(color='white', linewidth=2.5),
                whiskerprops=dict(linewidth=1.3),
                capprops=dict(linewidth=1.3),
                flierprops=dict(marker='o', markersize=3.5, alpha=0.4))

for patch, col in zip(bp['boxes'], exp_colors):
    patch.set_facecolor(col)
    patch.set_alpha(0.80)

# Overlay individual data points with horizontal jitter
for i, (data, col) in enumerate(zip(exp_data, exp_colors), 1):
    x = np.random.normal(i, 0.07, size=len(data))
    ax.scatter(x, data / 1000, alpha=0.25, color=col, s=18, zorder=3)

# Median labels above boxes
for i, data in enumerate(exp_data, 1):
    med = np.median(data) / 1000
    ax.text(i, med + 5, f'${med:.0f}k', ha='center',
            fontsize=10, fontweight='bold', color='#333')

ax.set_xticklabels(exp_labels, fontsize=11)
ax.yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'${x:.0f}k'))
ax.set_ylabel('Annual Salary (USD)', fontsize=11)
ax.set_title('How Much Does Experience Add to Your AI Salary?\nSalary Distribution by Experience Level',
             fontsize=14, fontweight='bold', pad=15)

# Step-up annotations
medians = [np.median(d) / 1000 for d in exp_data]
for i in range(1, len(medians)):
    delta = medians[i] - medians[i-1]
    ax.annotate(f'+${delta:.0f}k',
                xy=(i + 0.5, (medians[i] + medians[i-1]) / 2),
                fontsize=9.5, color='#388E3C', fontweight='bold', ha='center',
                bbox=dict(boxstyle='round,pad=0.3', facecolor='#E8F5E9', edgecolor='#388E3C', lw=0.8))

ax.text(0.01, -0.07,
        'Source: AI Jobs Market 2025–2026 Dataset | Each dot = one job posting',
        transform=ax.transAxes, fontsize=7.5, color='#888', style='italic')

plt.tight_layout()
plt.savefig('results/ai_salary_by_experience.png', dpi=180, bbox_inches='tight', facecolor='#FAFAF8')
plt.show()
```

### 5.3 🗺️ Geographic Insights (Salary by Country)

#### What This Code Does:
This section filters out geographic-agnostic roles (labeled 'Global' remote) to focus on physical countries. It aggregates the median and mean salaries, counts the number of postings per country, and overlays:
- A **Bar Chart** representing the median salary.
- **Diamond Markers (`D`)** representing the mean salary.
- **Text badges** showing the median salary and sample size (`n=...`).

#### Why We Wrote It This Way:
- **Comparing Mean vs Median**: By overlaying both metrics, we show the skewness in compensation. If the mean (diamond) is higher than the median (bar), it indicates a small cluster of very high-earning outliers stretching the average.
- **Sample Size (`n`) Annotation**: Crucial in salary studies to check data credibility. Knowing a country has high pay across 2 postings vs 500 postings is vital context for statistical reliability.


```python
country_stats = (df[df['country'] != 'Global']
                   .groupby('country')
                   .agg(median_sal=('annual_salary_usd','median'),
                        mean_sal=('annual_salary_usd','mean'),
                        count=('job_id','count'))
                   .reset_index()
                   .sort_values('median_sal', ascending=False))

fig, ax = plt.subplots(figsize=(12, 7))
fig.patch.set_facecolor('#FAFAF8')
ax.set_facecolor('#FAFAF8')

colors_c = [PALETTE[i % len(PALETTE)] for i in range(len(country_stats))]
bars = ax.bar(country_stats['country'],
              country_stats['median_sal'] / 1000,
              color=colors_c, alpha=0.88,
              edgecolor='white', linewidth=0.8)

# Plot mean values as diamonds to observe distribution skewness
ax.scatter(range(len(country_stats)),
           country_stats['mean_sal'] / 1000,
           color='#333', s=55, zorder=5, label='Mean salary', marker='D')

# Label bar tops with median value and job posting sample size (n), positioned above both median and mean to avoid overlap
for bar, (_, row) in zip(bars, country_stats.iterrows()):
    y_pos = max(row["median_sal"], row["mean_sal"]) / 1000 + 4.5
    ax.text(bar.get_x() + bar.get_width() / 2,
            y_pos,
            f'${row["median_sal"]/1000:.0f}k\nn={int(row["count"])}',
            ha='center', va='bottom', fontsize=8, fontweight='bold', color='#333')

ax.yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'${x:.0f}k'))
ax.set_xlabel('Country', fontsize=11)
ax.set_ylabel('Median Annual Salary (USD)', fontsize=11)
ax.set_title('Where Do AI Professionals Earn the Most?\nMedian Salary by Country (2025–2026)',
             fontsize=14, fontweight='bold', pad=15)
ax.legend(fontsize=9, framealpha=0.9)
ax.text(0.01, -0.09,
        'Source: AI Jobs Market 2025–2026 Dataset | "Global" remote roles excluded',
        transform=ax.transAxes, fontsize=7.5, color='#888', style='italic')

plt.tight_layout()
plt.savefig('results/ai_salary_by_country.png', dpi=180, bbox_inches='tight', facecolor='#FAFAF8')
plt.show()
```

### 5.4 🏠 Work Mode Comparison (Remote vs. On-site vs. Hybrid)

#### What This Code Does:
This section builds a 3-panel subplot comparing salary distributions across work arrangement models: **On-site**, **Hybrid**, and **Fully Remote**. For each panel:
- A **Histogram** illustrates the frequency counts of salaries.
- A **Thick Dashed Line** represents the median.
- A **Thin Dotted Line** represents the mean.

#### Why We Wrote It This Way:
- **Grid Comparison**: Renders three distinct panels side-by-side using the exact same Y-axis (`sharey=True`) to allow instant comparison of volume and shape differences.
- **Reference Lines**: Overlaying mean and median reference lines highlights the skewness in distributions for each class.


```python
remote_order  = ['On-site', 'Hybrid', 'Fully Remote']
remote_colors = ['#F44336', '#FF9800', '#4CAF50']
remote_data   = [df[df['remote_work'] == r]['annual_salary_usd'].values for r in remote_order]

fig, axes = plt.subplots(1, 3, figsize=(14, 6), sharey=True)
fig.patch.set_facecolor('#FAFAF8')

for ax, rtype, col, data in zip(axes, remote_order, remote_colors, remote_data):
    ax.set_facecolor('#FAFAF8')
    ax.hist(data / 1000, bins=20, color=col, alpha=0.78,
            edgecolor='white', linewidth=0.5)
    med = np.median(data) / 1000
    mn  = np.mean(data) / 1000
    ax.axvline(med, color='#333', linewidth=2, linestyle='--', label=f'Median: ${med:.0f}k')
    ax.axvline(mn,  color='#999', linewidth=1.5, linestyle=':',  label=f'Mean: ${mn:.0f}k')
    ax.set_title(rtype, fontsize=13, fontweight='bold', color=col, pad=10)
    ax.set_xlabel('Annual Salary (USD)', fontsize=10)
    ax.xaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'${x:.0f}k'))
    ax.legend(fontsize=8.5, framealpha=0.9)
    ax.text(0.5, -0.14, f'n = {len(data)}  |  Median ${med:.0f}k',
            transform=ax.transAxes, ha='center',
            fontsize=9, fontweight='bold', color='#555')

axes[0].set_ylabel('Number of Job Postings', fontsize=11)
fig.suptitle('Does Remote Work Pay More?\nSalary Distribution by Work Arrangement (2025–2026)',
             fontsize=14, fontweight='bold', y=1.02)
fig.text(0.5, -0.06,
         'Source: AI Jobs Market 2025–2026 Dataset',
         ha='center', fontsize=7.5, color='#888', style='italic')

plt.tight_layout()
plt.savefig('results/ai_salary_remote_vs_onsite.png', dpi=180, bbox_inches='tight', facecolor='#FAFAF8')
plt.show()
```

### 5.5 🔑 In-Demand Skills & Associated Compensation

#### What This Code Does:
Since skills are recorded as pipe-separated lists (`Python|SQL|PyTorch`), this script parses and tokenizes the strings, counts the occurrences of each skill to identify the top 25, and computes the median salary of postings that require each skill. It renders:
- A **Horizontal Bar Chart**: Showing skill frequency (demand volume).
- A **Colormap (RdYlGn)**: Coloring bars by median salary (green = high salary, red = lower salary).
- A **Colorbar scale**: Matching color hues to salary values ($k).

#### Why We Wrote It This Way:
- **Integrating Two Dimensions (Frequency + Salary)**: Using bar length for popularity (volume) and bar color for compensation (value) packs massive exploratory insights into a single visual representation.
- **Substring Mapping**: We run `df['required_skills'].str.contains(skill, regex=False)` to correctly locate all postings requiring the skill.


```python
all_skills = []
for skills_str in df['required_skills']:
    all_skills.extend([s.strip() for s in skills_str.split('|')])

skill_counts = Counter(all_skills)
top_skills   = pd.DataFrame(skill_counts.most_common(25),
                             columns=['skill', 'count'])

skill_salaries = {}
for skill in top_skills['skill']:
    mask = df['required_skills'].str.contains(skill, regex=False)
    skill_salaries[skill] = df[mask]['annual_salary_usd'].median() / 1000
top_skills['median_salary_k'] = top_skills['skill'].map(skill_salaries)

top_skills = top_skills.sort_values('count', ascending=True)

fig, ax = plt.subplots(figsize=(13, 10))
fig.patch.set_facecolor('#FAFAF8')
ax.set_facecolor('#FAFAF8')

norm = plt.Normalize(top_skills['median_salary_k'].min(),
                     top_skills['median_salary_k'].max())
cmap = plt.cm.RdYlGn
colors_skill = cmap(norm(top_skills['median_salary_k'].values))

bars = ax.barh(top_skills['skill'], top_skills['count'],
               color=colors_skill, alpha=0.90,
               edgecolor='white', linewidth=0.6)

for bar, (_, row) in zip(bars, top_skills.iterrows()):
    ax.text(bar.get_width() + 2,
            bar.get_y() + bar.get_height() / 2,
            f"{int(row['count'])}  |  ${row['median_salary_k']:.0f}k",
            va='center', fontsize=8.5, color='#333', fontweight='bold')

sm = plt.cm.ScalarMappable(cmap=cmap, norm=norm)
sm.set_array([])
cbar = plt.colorbar(sm, ax=ax, shrink=0.6, pad=0.01)
cbar.set_label('Median Salary of Roles\nRequiring This Skill ($k)', fontsize=9)

ax.set_xlabel('Number of Job Postings', fontsize=11)
ax.set_title('Top 25 Most In-Demand AI Skills (2025–2026)\nColour = Median Salary of Roles Requiring That Skill',
             fontsize=14, fontweight='bold', pad=15)
ax.set_xlim(0, top_skills['count'].max() * 1.25)
ax.text(0.01, -0.05,
        'Source: AI Jobs Market 2025–2026 Dataset',
        transform=ax.transAxes, fontsize=7.5, color='#888', style='italic')

plt.tight_layout()
plt.savefig('results/ai_top_skills_demand_salary.png', dpi=180, bbox_inches='tight', facecolor='#FAFAF8')
plt.show()
```

### 5.6 🧬 The LLM Premium (LLM Roles vs. Traditional AI)

#### What This Code Does:
This compares Large Language Model (LLM) roles directly against general/traditional AI roles across five distinct metrics:
1. **Median Salary**: Compensation benchmark.
2. **AI Salary Premium %**: Premium compared to non-AI IT roles.
3. **Demand Score**: Marketplace popularity score.
4. **YoY Demand Growth %**: Growth acceleration rate.
5. **Benefits Score**: Quality of employee perks.
It renders a 5-panel comparative bar chart array, calculating and displaying the 'LLM Premium' percentage directly below each metric plot.

#### Why We Wrote It This Way:
- **Multi-dimensional Comparison**: Renders separate subplots for each metric to show that the LLM role premium extends beyond basic salary to growth and perks.
- **Premium Calculations**: Dynamically computes `((LLM_Val - Non_LLM_Val) / Non_LLM_Val) * 100` and displays it at the bottom to provide immediate numerical insights.


```python
llm     = df[df['is_llm_role'] == 1]
non_llm = df[df['is_llm_role'] == 0]

metrics = {
    'Median Salary ($k)':         [llm['annual_salary_usd'].median()/1000,
                                   non_llm['annual_salary_usd'].median()/1000],
    'AI Salary Premium (%)':      [llm['ai_salary_premium_pct'].mean(),
                                   non_llm['ai_salary_premium_pct'].mean()],
    'Demand Score (/100)':        [llm['demand_score'].mean(),
                                   non_llm['demand_score'].mean()],
    'YoY Demand Growth (%)':      [llm['demand_growth_yoy_pct'].mean(),
                                   non_llm['demand_growth_yoy_pct'].mean()],
    'Benefits Score (/10)':       [llm['benefits_score_10'].mean(),
                                   non_llm['benefits_score_10'].mean()],
}

fig, axes = plt.subplots(1, 5, figsize=(17, 6))
fig.patch.set_facecolor('#FAFAF8')

for ax, (metric, (llm_val, non_val)) in zip(axes, metrics.items()):
    ax.set_facecolor('#FAFAF8')
    bars = ax.bar(['LLM\nRoles', 'Non-LLM\nRoles'],
                  [llm_val, non_val],
                  color=['#7B1FA2', '#1976D2'], alpha=0.85,
                  edgecolor='white', linewidth=0.8,
                  width=0.5)
    for bar, val in zip(bars, [llm_val, non_val]):
        ax.text(bar.get_x() + bar.get_width()/2,
                bar.get_height() + max(llm_val, non_val) * 0.02,
                f'{val:.1f}', ha='center', fontsize=10,
                fontweight='bold', color='#333')
    delta = llm_val - non_val
    pct   = (delta / non_val) * 100
    sign  = '+' if delta > 0 else ''
    ax.set_title(metric, fontsize=9.5, fontweight='bold', pad=8)
    ax.text(0.5, -0.18, f'LLM premium: {sign}{pct:.1f}%',
            transform=ax.transAxes, ha='center',
            fontsize=8.5, color='#7B1FA2' if delta > 0 else '#E53935',
            fontweight='bold')
    ax.tick_params(axis='x', labelsize=9)
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.grid(axis='y', color='#E8E8E3', linewidth=0.6)

llm_patch     = mpatches.Patch(color='#7B1FA2', alpha=0.85, label=f'LLM Roles (n={len(llm)})')
non_llm_patch = mpatches.Patch(color='#1976D2', alpha=0.85, label=f'Non-LLM Roles (n={len(non_llm)})')
fig.legend(handles=[llm_patch, non_llm_patch], loc='lower center',
           ncol=2, fontsize=10, bbox_to_anchor=(0.5, -0.08), framealpha=0.9)

fig.suptitle('LLM Roles vs. Everything Else: Are They Worth the Hype?\nKey Metrics Comparison — AI Jobs 2025–2026',
             fontsize=14, fontweight='bold', y=1.04)
fig.text(0.5, -0.13,
         'Source: AI Jobs Market 2025–2026 Dataset',
         ha='center', fontsize=7.5, color='#888', style='italic')

plt.tight_layout()
plt.savefig('results/ai_llm_vs_nonllm_comparison.png', dpi=180, bbox_inches='tight', facecolor='#FAFAF8')
plt.show()
```

## 6. 🏁 Key Findings & Strategic Recommendations

### 🔑 Core Takeaways
1. **LLM Roles Command a Large Premium**: Specialized roles focusing on LLMs and Generative AI are highly sought after. They offer a significant premium in salary, benefits, and YoY demand growth compared to general AI roles.
2. **Experience is Highly Valued**: Moving from Entry-level to Senior and Lead roles yields massive compensation jumps, especially the transition from Mid-level to Senior level, which marks the highest increase.
3. **Geographic Hotspots**: Compensation varies significantly by region. Hubs like the USA and Singapore command higher median salaries compared to other markets.
4. **Remote and Hybrid Flexibility**: Fully remote and hybrid models do not suffer a salary discount; instead, they remain highly competitive with, or occasionally exceed, on-site salary offerings.
5. **High-Value Skills**: Core AI skills like **Python**, **Deep Learning**, **LLM APIs**, and **Vector Databases** are critical for securing high-paying roles.
